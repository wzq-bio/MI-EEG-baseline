```python
import mne
from mne.datasets import eegbci
files=eegbci.load_data(1,[4])
raw=mne.io.read_raw_edf(files[0],preload=True)
raw.plot()
```

```python
raw.filter(8,30)
raw.plot()
```

```python
%matplotlib qt
raw.plot()
```

```python
raw.plot(duration=5)
```

```python
events,event_id=mne.events_from_annotations(raw)
print("事件字典",event_id)
epochs=mne.Epochs(raw,events,event_id,tmin=-0.5,tmax=4.0,preload=True)
print(epochs)
epochs.plot()
```

```python
epochs[:5].plot(n_channels=10,n_epochs=5)
```

```python
epochs.average().plot()
```
